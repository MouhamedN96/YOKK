## Abstract

This notebook reports zero-shot retrieval performance for multilingual and code-switched settings. We evaluate three tracks: (1) FR/WO/ENG multilingual retrieval, (2) Waxal Fulani retrieval with plain and code-switched queries, and (3) a scenario-based bargain intent benchmark. Metrics are MRR@10, Recall@10, and nDCG@10.


## Research Questions

1. How does zero-shot retrieval quality vary across language styles (`english`, `french`, `wolof`, mixed code-switch)?
2. Does code-switching degrade retrieval on Waxal compared with plain queries?
3. In a bargain-intent scenario slice, does style (`plain` vs `codeswitch`) change retrieval quality?

## Experimental Protocol

- Ranking depth: `top_k = 10`
- Evaluation type: zero-shot retrieval (no supervised fine-tuning for reported runs)
- Core metrics:
  - MRR@10: quality of first relevant hit ranking
  - Recall@10: relevant coverage in top-10
  - nDCG@10: rank-sensitive gain quality

````python
# Optional: Load all results and plot quick comparison charts
import json
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

fr_path = Path(r"C:\Users\momo-\OneDrive\Desktop\YAATAL\artifacts\lfm_colbert_fr_en_wo\run-20260227-083651\metrics.json")
wx_path = Path(r"C:\Users\momo-\OneDrive\Desktop\YAATAL\artifacts\lfm_colbert_waxal\run-20260227-083844\metrics.json")
sc_path = Path(r"C:\Users\momo-\OneDrive\Desktop\YAATAL\artifacts\lfm_colbert_scenarios\bargain\run-20260227-085047\metrics.json")

fr = json.loads(fr_path.read_text(encoding='utf-8'))
wx = json.loads(wx_path.read_text(encoding='utf-8'))
sc = json.loads(sc_path.read_text(encoding='utf-8'))

rows = []
for style, m in fr['metrics_by_style'].items():
    rows.append({'track': 'fr_wo_eng', 'style': style, 'mrr_at_10': m['mrr_at_k'], 'recall_at_10': m['recall_at_k'], 'ndcg_at_10': m['ndcg_at_k']})
for style, m in wx['baseline_by_eval_style'].items():
    rows.append({'track': 'waxal', 'style': style, 'mrr_at_10': m['mrr_at_k'], 'recall_at_10': m['recall_at_k'], 'ndcg_at_10': m['ndcg_at_k']})
for style, m in sc['metrics_by_style'].items():
    rows.append({'track': 'bargain_scenario', 'style': style, 'mrr_at_10': m['mrr_at_k'], 'recall_at_10': m['recall_at_k'], 'ndcg_at_10': m['ndcg_at_k']})

df = pd.DataFrame(rows)
display(df.sort_values(['track','style']).reset_index(drop=True))

for metric in ['mrr_at_10', 'recall_at_10', 'ndcg_at_10']:
    pivot = df.pivot(index='style', columns='track', values=metric)
    ax = pivot.plot(kind='bar', figsize=(9,4), title=f'{metric} by style and track')
    ax.set_ylabel(metric)
    ax.set_xlabel('style')
    plt.tight_layout()
    plt.show()
````
